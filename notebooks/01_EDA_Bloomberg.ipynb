# Análisis Exploratorio de Datos (EDA)
Este notebook realiza un análisis exploratorio inicial de las variables correspondientes a los datasets extraídos de Bloomberg.
El objetivo es identificar la calidad de los datos, valores faltantes, distribuciones y correlaciones para la predicción del tipo de cambio USD/PEN.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import glob
import os

# Configuración de estilo
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('muted')
```

```python
# Listar todos los archivos Excel extraídos de Bloomberg
data_path = '../data/raw/'
excel_files = glob.glob(os.path.join(data_path, '*.xlsx'))

print(f'Se han encontrado {len(excel_files)} datasets en la carpeta data/raw.')

# Identificar el archivo más grande para una exploración inicial
largest_file = max(excel_files, key=os.path.getsize)
print(f'\nCargando archivo principal para exploración: {os.path.basename(largest_file)}')

df_sample = pd.read_excel(largest_file)
display(df_sample.head())
```
