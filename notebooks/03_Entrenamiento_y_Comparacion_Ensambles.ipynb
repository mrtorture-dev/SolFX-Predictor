Este notebook consolida y ejecuta los experimentos reproducibles de regresión directa para los cuatro algoritmos de ensamble del proyecto **SolFX-Predictor**:
1. **XGBoost** (`Entrenando_el_XGBoost.py`)
2. **LightGBM** (`Entrenando_el_LightGBM.py`)
3. **Random Forest** (`Entrenando_el_RandomForest.py`)
4. **CatBoost** (`Entrenando_el_CatBoost.py`)

### Diseño Experimental Común
- **Horizontes de predicción ($h$):** 1, 5 y 20 días hábiles.
- **Variable objetivo:** Retorno acumulado en puntos básicos $y_{t,h} = 10\,000 \times \ln(P_{t+h}/P_t)$.
- **Variables predictoras (18):** Indicadores técnicos causales (`shift(1)`) y codificación cíclica de calendario.
- **Validación temporal:** `TimeSeriesSplit(n_splits=3, test_size=252, gap=h)` sobre 8 462 registros de entrenamiento.
- **Evaluación fuera de muestra:** 60 sesiones de *holdout* con origen móvil (*rolling origin*) frente a Caminata Aleatoria.
